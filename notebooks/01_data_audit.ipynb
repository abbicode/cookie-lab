````python
exclude_terms = [
    "oatmeal",
    "peanut butter",
    "pumpkin",
    "banana",
    "bar",
    "cookie cake",
    "cookie pizza",
    "cookie dough",
    "raw",
    "gluten free",
    "gluten-free",
    "vegan",
    "eggless",
    "no eggs",
    "passover",
    "brownie",
    "white chocolate",
    "mint",
    "pudding",
    "cream cheese",
    "orange",
    "sour cream"
]
````

````python
name_lower = choc_chip_candidates["Name"].str.lower()

exclude_mask = False

for term in exclude_terms:
    exclude_mask = exclude_mask | name_lower.str.contains(
        term,
        regex=False,
        na=False
    )

classic_candidates = choc_chip_candidates[
    ~exclude_mask
].copy()

print("Before filtering:", len(choc_chip_candidates))
print("After filtering:", len(classic_candidates))
````

````python
classic_ids = classic_candidates["RecipeId"]

classic_reviews = reviews[
    reviews["RecipeId"].isin(classic_ids)
].copy()

print("Classic cookie recipes:", len(classic_candidates))
print("Associated reviews:", len(classic_reviews))
````

````python
review_counts = (
    classic_reviews
    .groupby("RecipeId")
    .size()
    .rename("actual_review_count")
)

classic_with_review_counts = classic_candidates.merge(
    review_counts,
    on="RecipeId",
    how="left"
)

classic_with_review_counts["actual_review_count"] = (
    classic_with_review_counts["actual_review_count"]
    .fillna(0)
    .astype(int)
)

classic_with_review_counts[
    ["Name", "ReviewCount", "actual_review_count"]
].sort_values(
    "actual_review_count",
    ascending=False
).head(20)
````

````python
phenotype_terms = [
    "chewy",
    "chewier",
    "crispy",
    "crisp",
    "crunchy",
    "soft",
    "gooey",
    "cakey",
    "cake-like",
    "thick",
    "thin",
    "flat",
    "spread",
    "spreading",
    "dry",
    "moist",
    "dense"
]

for term in phenotype_terms:

    mask = (
        classic_reviews["Review"]
        .fillna("")
        .str.contains(
            term,
            case=False,
            regex=False
        )
    )

    n_reviews = mask.sum()

    n_recipes = (
        classic_reviews
        .loc[mask, "RecipeId"]
        .nunique()
    )

    print(
        f"{term:12} "
        f"reviews={n_reviews:5,} | "
        f"recipes={n_recipes:4,}"
    )
````

````python
def sample_trait_reviews(trait, n=20):
    matches = classic_reviews[
        classic_reviews["Review"]
        .fillna("")
        .str.contains(
            trait,
            case=False,
            regex=False
        )
    ]

    print(f"Reviews containing '{trait}': {len(matches)}")

    return matches[
        ["RecipeId", "Rating", "Review"]
    ].sample(
        min(n, len(matches)),
        random_state=42
    )
````

## Goal

The objective of this analysis is to determine whether recipe reviews can be used to extract measurable cookie characteristics ("cookie phenotypes").

- Thickness
- Spread / Thinness
- Moisture

- Gooeyness

Reviews were analyzed to determine whether users naturally describe these characteristics and whether they provide enough signal for future NLP modeling.

# Classic Chocolate Chip Cookie Dataset Cleaning

## Objective

The goal of this project is to create a machine learning dataset that can predict chocolate chip cookie texture (such as soft, chewy, thin, or crispy) based on recipe characteristics including ingredients and instructions.

To create this dataset, the original recipe database was filtered to isolate classic chocolate chip cookie recipes and then combined with user reviews to generate texture labels.

---

# 1. Initial Dataset

The original dataset contained:

- **522,517 recipes**
- **1,401,982 reviews**

Each recipe contained information including:

- Recipe ID
- Recipe name
- Recipe category
- Ingredients
- Instructions
- Author information
- Recipe metadata

---

# 2. Identify Chocolate Chip Cookie Candidates

Recipes were first filtered based on recipe names containing chocolate chip cookie-related terms.

This created:

`choc_chip_candidates`

**1,775 recipes**

This dataset contained both true chocolate chip cookie recipes and unrelated recipes that included the phrase "chocolate chip cookie."

Examples of unrelated recipes included:

- Chocolate Chip Cookie Pie
- Chocolate Chip Cookie Ice Cream Cake
- Chocolate Chip Cookie Cheesecake
- Chocolate Chip Cookie Milkshake

---

# 3. Remove Non-Cookie Recipes

The candidate dataset was cleaned to remove recipes that were not traditional chocolate chip cookies.

Recipes were removed based on:

- Recipe categories
- Recipe names
- Keywords indicating non-cookie desserts

Examples of removed categories:

- Bar Cookie
- Pie
- Frozen Desserts
- Cheesecake
- Ice Cream
- Beverages
- Cakes
- Sandwiches
- Cookie Mixes

After removing unrelated recipes:

`classic_candidates`

**1,258 recipes**

---

# 4. Refine Classic Chocolate Chip Cookie Recipes

Additional cleaning was performed to create a more representative classic chocolate chip cookie dataset.

This produced:

`classic_cookies_clean`

**1,205 recipes**

Duplicate recipe names were then investigated.

A total of:

**421 duplicate recipe names**

were identified.

Duplicates were reviewed because identical recipe names could represent different recipes from different authors.

After duplicate analysis, the final cleaned recipe dataset was:

`classic_cookies_final`

**1,180 recipes**

This dataset contains:

- Recipe ID
- Recipe name
- Recipe category
- Ingredients
- Instructions
- Author information
- Recipe metadata

---

# 5. Merge Recipes With Reviews

The cleaned recipe dataset was merged with the review dataset using:

`RecipeId`

An inner join was used so that only recipes with available reviews were retained.

The resulting dataset:

`classic_cookie_reviews`

contained:

**5,515 review rows**

Each row represented:

- One review
- The associated cookie recipe
- Recipe information

---

# 6. Convert Review-Level Data Into Recipe-Level Data

Because each recipe could have many reviews, the review-level dataset was converted into a recipe-level dataset.

Reviews were grouped by recipe and combined.

For each recipe:

- All reviews were combined into one text field
- The number of reviews was calculated
- The average rating was calculated

The final recipe-level dataset:

`classic_cookie_recipe_level`

contained:

**594 recipes**

Each row represents one cookie recipe.

Schema:

| Column | Description |
|---|---|
| RecipeId | Unique recipe identifier |
| Name | Recipe name |
| RecipeIngredientParts | List of recipe ingredients |
| RecipeInstructions | Baking instructions |
| num_reviews | Number of reviews associated with the recipe |
| avg_rating | Average review rating |
| all_reviews | All reviews combined into one text field |

---

# 7. Final Dataset Purpose

The final dataset will be used to predict cookie texture.

The recipe information will serve as model inputs:

- Ingredients
- Instructions
- Recipe characteristics

The combined reviews will be used to generate texture labels.

Example:

Reviews:

> "These cookies are soft, chewy, and stay thick."

Converted into labels:

| Texture | Label |
|---|---|
| Soft | 1 |
| Chewy | 1 |
| Thick | 1 |
| Crispy | 0 |

These labels will become the prediction targets for machine learning models.
