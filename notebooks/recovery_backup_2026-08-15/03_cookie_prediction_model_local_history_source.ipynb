### Butter Creaming and Chewiness

Butter creaming showed only a small association with chewy texture. Cookies made with creamed butter were classified as chewy 39.5% of the time compared with 34.8% of cookies without creamed butter. However, a chi-square test found this relationship was not statistically significant (p = 0.323).

## Chewiness Takeaway

Chewiness was most strongly associated with **brown sugar content and baking time**, rather than the presence of individual fats or mixing methods.

Brown sugar was the strongest ingredient predictor of chewiness. Recipes containing brown sugar were classified as chewy **39.8%** of the time compared with only **14.3%** of recipes without brown sugar. This relationship was statistically significant (**χ² test, p = 0.00034**), supporting the role of molasses-containing sugars in increasing moisture retention and creating a softer, chewier texture.

Bake time was also a significant predictor. Chewy cookies had shorter average bake times compared with non-chewy cookies (**t-test, p = 0.00536**), consistent with baking science: shorter baking preserves moisture, while longer baking removes water and promotes crispness.

Other commonly discussed factors showed weaker effects in this dataset. Shortening, butter, cornstarch, and butter creaming showed some positive trends toward chewiness, but none were statistically significant. This suggests that while these ingredients may influence texture, their effects were smaller than the combined impact of sugar composition and baking conditions.

Overall, the model and statistical analysis suggest that **a chewy cookie is primarily created by retaining moisture: using brown sugar and avoiding excessive baking time are the strongest predictors of a chewy texture.**

## Thickness Analysis

Thickness was analyzed to determine whether ingredient composition or recipe characteristics contributed to producing thicker cookies.

Ingredient ratios showed limited evidence of influencing thickness. Fat/flour ratio and brown sugar fraction were not significantly associated with thickness (p > 0.05). Sugar/flour ratio showed the strongest trend among the tested ratios, with thick cookies having a slightly different sugar-to-flour balance, but this relationship was still not statistically significant (p = 0.113). This suggests that overall ingredient proportions alone do not strongly predict cookie thickness.

Since thickness is closely related to cookie shape and spread behavior, structural characteristics were also considered. Unlike crispiness, which was strongly associated with thinness, flatness, and spreading, thickness likely reflects reduced spreading and greater dough structure. This suggests that physical characteristics of the cookie may play a larger role in determining thickness than individual ingredient ratios.

Overall, thickness did not appear to be driven by a single ingredient ratio. Instead, thickness is likely influenced by a combination of structural factors that affect how much a cookie spreads during baking.
