# Model Evaluation & Limitations

## Overview

After analyzing cookie texture using statistical testing and machine learning models, this section evaluates model performance, identifies limitations in the dataset, and discusses opportunities for improving future predictions.

The goal of the models was not only to maximize accuracy, but also to understand which aspects of cookie texture can be predicted from recipe information and which require additional measurements.

---

# Model Performance Summary

Two classification models were compared:

- **Logistic Regression**
  - Provides interpretable relationships between recipe features and texture outcomes.
  - Helps identify which variables increase or decrease the likelihood of a texture.

- **Random Forest**
  - Captures nonlinear relationships and interactions between ingredients and baking conditions.
  - Provides feature importance rankings.

## Results

- **Chewy**
  - Logistic Regression: **74.9% accuracy**
  - Random Forest: **87.0% accuracy**
  - Chewiness was the most predictable texture, suggesting recipe composition contains strong signals related to moisture retention and baking behavior.

- **Structural Crispy**
  - Logistic Regression: **48.4% accuracy**
  - Random Forest: **61.1% accuracy**
  - Crispiness was more difficult to predict from ingredients alone, suggesting physical characteristics such as spreading, thinness, and surface area are important drivers.

- **Soft**
  - Logistic Regression: **65.3% accuracy**
  - Random Forest: **61.1% accuracy**
  - Softness was difficult to predict, suggesting it is influenced by many interacting factors rather than a single dominant variable.

- **Thick**
  - Logistic Regression: **92.6% accuracy**
  - Random Forest: **87.4% accuracy**
  - These values were affected by class imbalance, as very few thick cookies were present. Both models struggled to identify thick cookies.

# Key Model Insights

## Recipe Chemistry vs. Physical Structure

The model results suggest that different textures have different underlying mechanisms.

- Chewiness was strongly influenced by recipe and baking variables.
- Crispiness was more dependent on physical structure, including:
  - thinness
  - spreading
  - flatness
- Softness and thickness likely require additional variables beyond recipe composition.

This demonstrates that ingredient-level data alone cannot fully capture the complexity of cookie texture.

---

# Dataset Limitations

## 1. Subjective Texture Labels

Texture labels were obtained from recipe descriptions and reviews, meaning classifications reflect human perception.

Terms such as:
- chewy
- soft
- crispy
- thick

may have different meanings between users.

For example, one person may describe a cookie as "soft" while another may describe the same cookie as "chewy."

This introduces noise into the target variables and limits model performance.

---

## 2. Lack of Standardized Recipe Measurements

Recipes varied in factors that strongly affect cookie outcomes, including:

- flour type and protein content
- measuring methods
- ingredient temperature
- cookie size
- baking sheet material
- oven calibration
- exact chilling time

Because these variables were not consistently available, the model could not account for all sources of variation.

---

## 3. Observational Data

The dataset contains existing recipes rather than controlled experiments.

Therefore, the analysis identifies **relationships and patterns**, but does not prove causation.

For example:

- Brown sugar may be associated with chewy cookies,
- but this does not prove that adding brown sugar alone will always create a chewy cookie.

Controlled experiments would be needed to determine causal effects.
